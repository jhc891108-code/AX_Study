2.리스트(List)

- 변경 가능
- [값1 값2,...] 

#### 주요 기능
추가 (create)
- append(요소): 끝에 추가
- insert(추가 위치, 추가 요소)
- extend(ruits콜렉션 - 리스트, 듀풀): 끝에 여러개 추가
- + [기존 리스트] [값1, 값2 ...] : 끝ㅌ에 여러개 추가

조회(Read)
- 인덱스 번호로 접근 (fruits[1])
- index(요소): 특정 요소의 인덱스 번호

수정 (update)
- 인덱스 번호로 접근, 다른 값을 대립 (fruits[1] = '오랜지')

삭제(Dplete)
- remove(요소) : 첫번째 발견한 특정 요소를 제거
- clear(): 리스트의 모든 요소 제거


기타
- sort() : 정렬(오름차순), sort(reverse=True) : 내림차순
- reverse() : 순서 반전
- count(요소) : 리스트의 모든 요소 제거

#### 논리 연산자(and, or, not)
and
-각 향이 모두 True -> True, 하나라도 False -> False

or

not
- True 조건 -> False 조건으로 반전, False 조건 -> Trune 조건으로 반전

```python
# and
num1 = 20

result1 = num1 > 10
result2 = num2 < 100

result1 and result2
```

1. 1세대: %연산자 방식

```python
- %s: 문자열 대체
    -%숫자s -> 최소 숫자 자리수 만큼 공간을 차지, 남는 공간은 공백
- %d: 정수
    %대체문자숫자s -> %05d -> 남은 공백은 0으로 대체
- %f: 실수
    -%.2f : thtnwja 2자리까지 표시
```

```python
message = "나의 이름은 %10s이고, 나이는 %d입니다."

message % ("철수", 20)
```

```python
message = "주문번호: %05d" % 10
message
```

```python
message = "숫자: %f" % 10 
```

2. 파이썬의 변수 검색 규칙:LEGB 스코프

접근 우선순위: 좁은 범위가 우선순위가 높다!

L (Local - 지역 변수) > E (Enclosing - 외부 함수 변수) > G (Global - 전역 변수) > B (Built-in - 내장 변수)

```python
a = 'global'
def outer():
    print("outer:", a)
    
    def inner():
        a = "locals"
        print("inner:", a)
        
    inner()

outer()
```

3. global 키워드

```python
a = 'Global'
def outer():
    a = 'Enclosing'
    print("outer:", a)
    
    def inner():
        print("inner:", a) # a라는 변수는 전역변수를 찾겠다.
        
    inner()

outer()
```

```python
a = True
def outer():
    a = 'Enclosing'
    print("outer:", a)
    
    def inner():
        global a # a라는 변수는 전역변수를 찾겠다.
        a = not a
        print("inner:", a)
        
    inner()

outer()
```

5. lambda 함수

- 간단하게 일회용으로 사용하는 경우 한줄의 식으로 간당하게 함수 정의
- 일회용으로 사용되므로 매개변수 이름도 최대한 간결하게(한글)

'''
lambda 매개변수, 매계변수, ...: 처리 부분
'''

```python
square = lambda num: num ** 2
square(10)
```

```python
def change_numbers(callback, *args):
    numbers = list(args)
    
    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])
        
    return numbers

change_numbers(lambda num: num ** 2, 10, 20, 30)
```

```python
change_numbers(lambda x: x ** 2, 10, 20, 30)
```

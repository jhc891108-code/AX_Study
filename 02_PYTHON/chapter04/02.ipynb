1. for 문

#### 횟수가 정해져 있는 반목
''''
for num in range(반복횟수) :
    # 반복 수행되는 코드
....

i : 순서를 의미하는 index, 관례, 다음 for문은 알파벳 순서대로 j, k, l, ..

```python
total = 0
start = 1
for num in range(100): # 100번 반복
    total = total + start
    total += start
    start += 1
    print("반복 횟수:", num)

print(f"1~100 합계: {total}")
```

2. 이터레이터(lterator)

반복을 하려면
- 이터레이터 객체 생성(\_\_iter_호출(iter(...))
- 순서를 한칸씩 이동하면서 요소 조회(\_\___next\_\_ 호출(next(..)))

```python
fruits = ['Apple', 'orang', 'Melon', 'Banana']

it = iter(fruits) #_iter_(..)호출
```

```python
next(it)
```

3. 제너레이터(Generator)

- 함수의 실행을 정지하고 월할 떄 다시 재개하는 형태

```python
def generate_nums():

    print("1번쨰")
    yield 10

    print("2번째")
    yield 20

    print("3번쨰")
    yield 30
```

```python
gen = generate_nums()
gen
```

```python
next(gen)
```

```python
next(gen)  # 값을 하나씩 반환하며 이전 실행 위치를 유지
```

4장 1장 : 조건문

- 판별식에 따라서 참(True)면 참에 실행되는 코드를 정의하고, 거짓(False)이면 거짓 조건에 실행되는 코드를 정의할 수 있는 문장!
- 판별식 : 비교 연산자, 논리 연산자, 값

1. 단일 조건문 if

'''
if 판별식:
    판별식이 참일떄 실행되는 코드
'''

```python
num = 10
if num > 5:
    print("5보다 크다")
```

```python
num = 5
if num == 10:
    print("10 입니다.")
else: # if num == 10: 거짓일떄 실행되는 코드를 정의
    print("10이 아닙니다.")
```

3. 단일 조건 분기: if-elif-else

```python
age = int(input())

if age >= 20:
    print("성인")
elif age >= 17:
    print("고등학생")
elif age >= 14:
    print("중학생")
elif age >= 8:
    print("초등학생")
else:
    print("유치원생")
```

4. 기본값 매개변수(Default Parameter)

매개변에 값을 할당하지 않아도 기본적으로 가질 수 있는 값 설정
- 기본값이 있는 매개변수는 오른쪽 끝부터 차례대로 정의

````python
def add(num1, num2, name, mobile = "010"):
    return f"{num1 + num2}, {name=}, {mobile=}"
````

````python
add(10, 20, "김영희")
````

5.가변 인수(*args)와 키워드 가변 인수(*kwargs)

- *args: 가변적인 위치 인수, 매치변수 갯수 상관없이 지정 가능, 키워드 배개변수로는 사용 불과

````python
def add(num1, num2, num3, num4):
    return num1 + num2 + num3 + num4

add(10, 20, 30, 40)
````

````python
def add(*nums):
    print(nums, type(nums))
````

````python
def add(*nums):
    #print(nums, type(nums))
    total = 0
    for num in nums:
        total += num
    return total
````

````python
print(add(10, 20))
print(add(10, 20, 30))
print(add(10, 20, 30, 40))
````

````python
order_database = [
    {"name": "김철수", "completed": True, "price": 60000},
    {"name": "이영희", "completed": False, "price": 35000},
    {"name": "박민수", "completed": True, "price": 25000},
    {"name": "최지영", "completed": True, "price": 85000},
    {"name": "정성민", "completed": False, "price": 90000}
]

total_price = 0
order_count = 0

for order in order_database:
    if order["completed"] == True:
        if order["price"] >= 50000:
            total_price += order["price"]
            order_count += 1

print(f"우수 고객 주문 건수: {order_count}건")
print(f"우수 고객 총 결제 금액: {total_price}원")
````

````python
fruit_sweetness = [12.5, 14.2, 11.8, 10.5, 15.0, 13.6, 11.2, 14.8, 12.0, 13.1]

max_sweetness = fruit_sweetness[0]
min_sweetness = fruit_sweetness[0]
total = 0

for sweetness in fruit_sweetness:
    if sweetness > max_sweetness:
        max_sweetness = sweetness
    if sweetness < min_sweetness:
        min_sweetness = sweetness
    total += sweetness

avg_sweetness = total / len(fruit_sweetness)

print(f"최고 과일 당도: {max_sweetness}")
print(f"최저 과일 당도: {min_sweetness}")
print(f"과일 평균 당도: {avg_sweetness:.2f}")
````
